# Books to Scrape: Simple Web Scraper | نمونهٔ خزش وب‌سایت Books to Scrape

Fetch and parse up to ten book titles and prices from the Books to Scrape demo site, saving CSV.

Open the repository README for prerequisites, input assets, limitations, and run instructions. | برای پیش‌نیازها، ورودی‌ها و راهنمای اجرا README مخزن را ببینید.

```python
import requests
from bs4 import BeautifulSoup
import csv

BASE_URL = "https://books.toscrape.com/"

response = requests.get(BASE_URL)
soup = BeautifulSoup(response.text, "html.parser")

books = soup.select("article.product_pod")[:10]

data = []
for b in books:
    title = b.h3.a["title"]
    price = b.select_one("p.price_color").text.strip()
    data.append({"title": title, "price": price})

with open("books_bs4.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.DictWriter(f, fieldnames=["title", "price"])
    writer.writeheader()
    writer.writerows(data)

print("Saved books_bs4.csv with 10 books.")
```
